# Matrix Product in Numpy

## Taking the Product of the Matrixes

```python
A = np.arange(15).reshape(3,5)
B = np.arange(15).reshape(5,3)
C = A @ B # Multiplies the matrixes
C
```

```python
# Element wise product of the matrixes
A.dot(B)
```

```python
A.min()
A.max()
```
