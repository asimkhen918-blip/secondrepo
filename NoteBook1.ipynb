## Basic Data Structures in the Python

# Tuples
## Ordered Collection of Elements
- Enclosed in () round braces / paranthesis
- Different kind of elements can be stored
- Once elements are stored you can't change them (Unmutable)

```python
tuple1 = (1, "Python", 23.5)
type(tuple1)
```

### Indexing in Tuple

```python
tuple1[1]
tuple1[0:3] # Large element is exclusive
```

```python
# Count of elements 
len(tuple1)
```

```python
tuple1 = (1, "Python", 23.5)  # Concatenation
tuple2 = (1, "Python", 23.5)
tuple1 + tuple2
tuple1*2 + tuple2*3   # Concateante and repeat
```

```python
tup = (12,23,45,67,89)
min(tup)
max(tup)
```

# Lists
- Ordered collection of elements 
- Enclosed in [ ] Square braces
- Mutable , you can change the values


```python
list1 = [2, "Asim", 23.4]
type(list1)
list2 = ["ali", 234, False, 23.4]
# Concatination
list1 + list2
```

```python
list1.reverse()
list1
```

```python
list1.append("Asim")
list1
```

```python
list3 = [23,20,10,40,45,89,23]
list3.sort() # sorting the list
list3
list1 + list3
```

# Dictionaries
- Unordered collection of elements
- Key and value
- Curly braces or brackets {}
- Mutateable / Chnage the values


```python
d1 = {"Samosa" : 30, "Pakora" : 40, "Raita": 50,"Salad": 20} # Unoredered Collection
d1.get("Samosa")
d1.__len__()
type(d1)
d1.keys()
d1.values()
d1["Samosa"] = 222  # Update
d1
```

```python
d2 = {"Dates":50, "Chocolates":200, "Sewayyian": 1000}
d1.update(d2)
d1
```
