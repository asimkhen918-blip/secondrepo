# Customizing the Plots in Python

```python
import seaborn as sns
import pandas as pd
import numpy as np

kashti = sns.load_dataset("titanic")
kashti.head(7)
```

```python
# 1. Import ALL libraries at the very top
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Load the dataset (assuming 'kashti' is the seaborn titanic dataset)
kashti = sns.load_dataset("titanic")

# 3. Create the plot
sns.boxplot(
    x="survived", 
    y="age",
    showmeans=True, 
    meanprops={"marker":"*", "markersize":"12", "markeredgecolor":"red"}, 
    data=kashti
)

# 4. Add labels and title (Notice: no commas at the end of these lines)
plt.xlabel("How many survived",size=10, weight="bold")
plt.ylabel("Age (Years)")
plt.title("Box plot of kitnay doobay, kitne tar gaye")

# 5. Display the plot
plt.show()
```
